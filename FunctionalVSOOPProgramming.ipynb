Sort 0s, 1s and 2s

```python
class Solution:
    def sort012(self, arr):
        low = 0
        mid = 0
        high = len(arr) - 1

        while mid <= high:
            if arr[mid] == 0:
                arr[low], arr[mid] = arr[mid], arr[low]
                low += 1
                mid += 1
            elif arr[mid] == 1:
                mid += 1
            else:  # arr[mid] == 2
                arr[mid], arr[high] = arr[high], arr[mid]
                high -= 1

        return arr
```

Binary Search

```python
class Solution:
    def firstSearch(self, arr, k):
        left = 0
        right = len(arr) - 1
        result = -1  # Stores the smallest index found so far

        while left <= right:
            mid = left + (right - left) // 2

            if arr[mid] == k:
                result = mid       # Record this index
                right = mid - 1    # Keep searching to the left
            elif arr[mid] < k:
                left = mid + 1
            else:
                right = mid - 1

        return result
```
