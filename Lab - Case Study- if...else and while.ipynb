```python
def main():
    while True:
        last_name = input("Enter student's last name (or 'ZZZ' to quit): ").strip()
       
        if last_name.upper() == 'ZZZ':
            print("Exiting Program")
            break

        first_name = input("Enter student's first name: ").strip()

        try:
            gpa = float(input("Enter the student's GPA: "))
        except ValueError:
            print("please inter a valid GPA number.\n")
            continue


        if gpa >= 3.5:
            print(f"Congratulations, {first_name} {last_name} you made the Dean's List!")
        elif gpa >= 3.25:
            print(f"Congratulations, {first_name} {last_name} you made the Honor Roll!")
        else:
            print(f"Sorry {first_name} {last_name} you do not qualify for the Dean's List or Honor Roll.")

        print()

if __name__ == "__main__":
    main()
```
